```python
db_path_3 = Path.cwd().parent / "data" / "articles.db"
connection = sqlite3.connect(db_path_3)

query_3 = "SELECT COUNT(source_site) as Number_of_articles, source_site as Name_of_site FROM articles GROUP BY source_site"

df = pd.read_sql_query(query_3, connection)

connection.close()

df
```
